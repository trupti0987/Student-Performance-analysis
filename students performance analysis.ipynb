```python
# ============================================================
# STUDENT PERFORMANCE ANALYSIS AND PREDICTION
# Complete Data Science Project
# ============================================================


# ============================================================
# 1. INSTALL AND IMPORT LIBRARIES
# ============================================================

!pip install ucimlrepo -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

print("Libraries loaded successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

print("Loading dataset...")

student_data = fetch_ucirepo(id=320)

# Get features
features = student_data.data.features.copy()

# Get target
target = student_data.data.targets.copy()

# Combine features and target
df = pd.concat([features, target], axis=1)

# Clean column names
df.columns = df.columns.astype(str).str.strip()

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. CHECK COLUMN NAMES
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================================
# 4. BASIC INFORMATION
# ============================================================

print("\nDataset Information:")
print(df.info())

print("\nDataset Description:")
display(df.describe())


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum().sum())

# Fill missing numerical values
for column in df.select_dtypes(include=np.number).columns:
    df[column] = df[column].fillna(df[column].mean())

print("Missing values handled!")


# ============================================================
# 6. REMOVE DUPLICATES
# ============================================================

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Dataset Shape after removing duplicates:", df.shape)


# ============================================================
# 7. CHECK TARGET COLUMN
# ============================================================

# The UCI Student Performance dataset has G3 as final grade.

if "G3" not in df.columns:

    # Find the column that represents G3
    possible_columns = []

    for column in df.columns:
        if str(column).strip().upper() == "G3":
            possible_columns.append(column)

    if len(possible_columns) > 0:
        df["G3"] = df[possible_columns[0]]

    else:
        print("Available columns:")
        print(df.columns.tolist())
        raise ValueError("G3 column was not found.")

print("\nTarget column G3 found successfully!")


# ============================================================
# 8. FINAL GRADE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))

sns.histplot(
    df["G3"],
    bins=20,
    kde=True
)

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()


# ============================================================
# 9. STUDENT COUNT BY GENDER
# ============================================================

if "sex" in df.columns:

    plt.figure(figsize=(6,5))

    sns.countplot(
        x="sex",
        data=df
    )

    plt.title("Number of Students by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Number of Students")

    plt.show()


# ============================================================
# 10. FINAL GRADE BY GENDER
# ============================================================

if "sex" in df.columns:

    plt.figure(figsize=(7,5))

    sns.boxplot(
        x="sex",
        y="G3",
        data=df
    )

    plt.title("Final Grade by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Final Grade")

    plt.show()


# ============================================================
# 11. STUDY TIME VS FINAL GRADE
# ============================================================

if "studytime" in df.columns:

    plt.figure(figsize=(8,5))

    sns.boxplot(
        x="studytime",
        y="G3",
        data=df
    )

    plt.title("Study Time vs Final Grade")
    plt.xlabel("Study Time")
    plt.ylabel("Final Grade")

    plt.show()


# ============================================================
# 12. ABSENCES VS FINAL GRADE
# ============================================================

if "absences" in df.columns:

    plt.figure(figsize=(8,5))

    plt.scatter(
        df["absences"],
        df["G3"],
        alpha=0.6
    )

    plt.xlabel("Number of Absences")
    plt.ylabel("Final Grade")
    plt.title("Absences vs Final Grade")

    plt.show()


# ============================================================
# 13. G1 VS G3
# ============================================================

if "G1" in df.columns:

    plt.figure(figsize=(8,5))

    plt.scatter(
        df["G1"],
        df["G3"],
        alpha=0.6
    )

    plt.xlabel("First Period Grade (G1)")
    plt.ylabel("Final Grade (G3)")
    plt.title("G1 vs G3")

    plt.show()


# ============================================================
# 14. G2 VS G3
# ============================================================

if "G2" in df.columns:

    plt.figure(figsize=(8,5))

    plt.scatter(
        df["G2"],
        df["G3"],
        alpha=0.6
    )

    plt.xlabel("Second Period Grade (G2)")
    plt.ylabel("Final Grade (G3)")
    plt.title("G2 vs G3")

    plt.show()


# ============================================================
# 15. INTERNET VS FINAL GRADE
# ============================================================

if "internet" in df.columns:

    plt.figure(figsize=(7,5))

    sns.boxplot(
        x="internet",
        y="G3",
        data=df
    )

    plt.title("Internet Access vs Final Grade")
    plt.xlabel("Internet Access")
    plt.ylabel("Final Grade")

    plt.show()


# ============================================================
# 16. CORRELATION ANALYSIS
# ============================================================

numeric_data = df.select_dtypes(include=np.number)

correlation = numeric_data.corr()

print("\nCorrelation with Final Grade:")
print(
    correlation["G3"].sort_values(
        ascending=False
    )
)


# ============================================================
# 17. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(14,10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 18. SELECT FEATURES
# ============================================================

# Select useful numerical columns.
# Only use columns that actually exist.

possible_features = [
    "age",
    "Medu",
    "Fedu",
    "traveltime",
    "studytime",
    "failures",
    "famrel",
    "freetime",
    "goout",
    "Dalc",
    "Walc",
    "health",
    "absences",
    "G1",
    "G2"
]

features_used = []

for column in possible_features:

    if column in df.columns:
        features_used.append(column)

print("\nFeatures used:")
print(features_used)


# ============================================================
# 19. CREATE X AND Y
# ============================================================

X = df[features_used].copy()

y = df["G3"].copy()

print("\nX Shape:", X.shape)
print("Y Shape:", y.shape)


# ============================================================
# 20. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))


# ============================================================
# 21. LINEAR REGRESSION
# ============================================================

linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)

print("\nLinear Regression trained successfully!")


# ============================================================
# 22. LINEAR REGRESSION PREDICTION
# ============================================================

linear_prediction = linear_model.predict(
    X_test
)


# ============================================================
# 23. LINEAR REGRESSION RESULTS
# ============================================================

linear_mae = mean_absolute_error(
    y_test,
    linear_prediction
)

linear_mse = mean_squared_error(
    y_test,
    linear_prediction
)

linear_rmse = np.sqrt(
    linear_mse
)

linear_r2 = r2_score(
    y_test,
    linear_prediction
)

print("\n================================")
print("LINEAR REGRESSION RESULTS")
print("================================")

print("MAE :", round(linear_mae, 2))
print("MSE :", round(linear_mse, 2))
print("RMSE:", round(linear_rmse, 2))
print("R2 Score:", round(linear_r2, 2))


# ============================================================
# 24. RANDOM FOREST
# ============================================================

random_forest = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

random_forest.fit(
    X_train,
    y_train
)

print("\nRandom Forest trained successfully!")


# ============================================================
# 25. RANDOM FOREST PREDICTION
# ============================================================

rf_prediction = random_forest.predict(
    X_test
)


# ============================================================
# 26. RANDOM FOREST RESULTS
# ============================================================

rf_mae = mean_absolute_error(
    y_test,
    rf_prediction
)

rf_mse = mean_squared_error(
    y_test,
    rf_prediction
)

rf_rmse = np.sqrt(
    rf_mse
)

rf_r2 = r2_score(
    y_test,
    rf_prediction
)

print("\n================================")
print("RANDOM FOREST RESULTS")
print("================================")

print("MAE :", round(rf_mae, 2))
print("MSE :", round(rf_mse, 2))
print("RMSE:", round(rf_rmse, 2))
print("R2 Score:", round(rf_r2, 2))


# ============================================================
# 27. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Random Forest"
    ],

    "MAE": [
        linear_mae,
        rf_mae
    ],

    "RMSE": [
        linear_rmse,
        rf_rmse
    ],

    "R2 Score": [
        linear_r2,
        rf_r2
    ]

})

print("\nModel Comparison:")

display(comparison)


# ============================================================
# 28. ACTUAL VS PREDICTED
# ============================================================

results = pd.DataFrame({

    "Actual Grade": y_test.values,

    "Linear Regression": linear_prediction,

    "Random Forest": rf_prediction

})

print("\nActual and Predicted Values:")

display(
    results.head(15)
)


# ============================================================
# 29. ACTUAL VS PREDICTED GRAPH
# ============================================================

plt.figure(figsize=(8,6))

plt.scatter(
    y_test,
    rf_prediction,
    alpha=0.7
)

plt.xlabel("Actual Grade")
plt.ylabel("Predicted Grade")

plt.title(
    "Actual vs Predicted Final Grade"
)

plt.show()


# ============================================================
# 30. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({

    "Feature": features_used,

    "Importance":
    random_forest.feature_importances_

})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFeature Importance:")

display(importance)


# ============================================================
# 31. FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10,6))

sns.barplot(
    x="Importance",
    y="Feature",
    data=importance
)

plt.title(
    "Feature Importance"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()


# ============================================================
# 32. PREDICT A NEW STUDENT
# ============================================================

# Create a new student using average values
# This makes the prediction code safe and simple.

new_student = pd.DataFrame(
    [X.mean().values],
    columns=X.columns
)

# Change some values for demonstration

if "studytime" in new_student.columns:
    new_student["studytime"] = 3

if "failures" in new_student.columns:
    new_student["failures"] = 0

if "absences" in new_student.columns:
    new_student["absences"] = 4

if "G1" in new_student.columns:
    new_student["G1"] = 14

if "G2" in new_student.columns:
    new_student["G2"] = 15


print("\nNew Student Data:")

display(new_student)


# ============================================================
# 33. PREDICT NEW STUDENT GRADE
# ============================================================

new_prediction = random_forest.predict(
    new_student
)

predicted_grade = new_prediction[0]

print("\n================================")
print("NEW STUDENT PREDICTION")
print("================================")

print(
    "Predicted Final Grade:",
    round(predicted_grade, 2)
)


# ============================================================
# 34. PERFORMANCE CATEGORY
# ============================================================

if predicted_grade >= 16:

    performance = "Excellent"

elif predicted_grade >= 12:

    performance = "Good"

elif predicted_grade >= 10:

    performance = "Average"

else:

    performance = "Needs Improvement"


print(
    "Performance:",
    performance
)


# ============================================================
# 35. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "student_performance_cleaned.csv",
    index=False
)

print(
    "\nCleaned dataset saved successfully!"
)


# ============================================================
# 36. SAVE PREDICTION RESULTS
# ============================================================

results.to_csv(
    "student_prediction_results.csv",
    index=False
)

print(
    "Prediction results saved successfully!"
)


# ============================================================
# 37. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("================================================")
print("       STUDENT PERFORMANCE ANALYSIS")
print("================================================")

print(
    "Total Students:",
    len(df)
)

print(
    "Average Final Grade:",
    round(df["G3"].mean(), 2)
)

print(
    "Highest Final Grade:",
    df["G3"].max()
)

print(
    "Lowest Final Grade:",
    df["G3"].min()
)

print(
    "Linear Regression R2:",
    round(linear_r2, 2)
)

print(
    "Random Forest R2:",
    round(rf_r2, 2)
)

print(
    "New Student Predicted Grade:",
    round(predicted_grade, 2)
)

print(
    "Performance:",
    performance
)
```
